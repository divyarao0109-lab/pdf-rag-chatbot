```python
from rag import extract_pages, chunk_pages

pages = extract_pages("your_file.pdf")   # put a PDF in the project folder
chunks = chunk_pages(pages, chunk_size=1000, chunk_overlap=200)
print(len(pages), "pages,", len(chunks), "chunks")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
texts = [c["text"] for c in chunks]
emb = model.encode(texts, normalize_embeddings=True)
print(emb.shape)   # (number of chunks, 384)
```

```python
import numpy as np

question = "Write a question your PDF can answer"
q_emb = model.encode([question], normalize_embeddings=True)[0]
scores = emb @ q_emb                       # similarity to every chunk
top = np.argsort(scores)[::-1][:3]         # best 3

for i in top:
    print(round(float(scores[i]), 3), "| page", chunks[i]["page"])
    print(chunks[i]["text"][:300])
    print("---")
```
